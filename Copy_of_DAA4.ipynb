```python
N = 8

board = [[0 for i in range(N)] for j in range(N)]

print("Initial Chessboard:")
for row in board:
    print(row)
```

```python
# Place the first Queen at row 0, column 0

board[0][0] = 1

print("Chessboard after placing first Queen:")
for row in board:
    print(row)
```

```python
def is_safe(board, row, col):

    # Check the same column
    for i in range(row):
        if board[i][col] == 1:
            return False

    # Check upper-left diagonal
    i = row - 1
    j = col - 1

    while i >= 0 and j >= 0:
        if board[i][j] == 1:
            return False
        i -= 1
        j -= 1

    # Check upper-right diagonal
    i = row - 1
    j = col + 1

    while i >= 0 and j < N:
        if board[i][j] == 1:
            return False
        i -= 1
        j += 1

    return True
```

```python
def solve_n_queens(board, row):

    # All Queens are successfully placed
    if row == N:
        return True

    # Try every column in the current row
    for col in range(N):

        if is_safe(board, row, col):

            # Place Queen
            board[row][col] = 1

            # Move to the next row
            if solve_n_queens(board, row + 1):
                return True

            # Backtrack if solution is not found
            board[row][col] = 0

    return False
```

```python
# First Queen is already placed at (0, 0)

if solve_n_queens(board, 1):

    print("Final N-Queens Matrix:\n")

    for row in board:
        print(row)

else:
    print("Solution does not exist.")
```
