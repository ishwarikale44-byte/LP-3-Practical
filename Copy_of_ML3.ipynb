```python
# Gradient Descent Algorithm

import numpy as np
import matplotlib.pyplot as plt

# Define Function
def function(x):
    return (x + 3)**2

# Define Derivative
def derivative(x):
    return 2 * (x + 3)

# Initialize Values
x = 2
learning_rate = 0.1
iterations = 20

# Store points for graph
x_points = [x]
y_points = [function(x)]

print("Gradient Descent Algorithm")
print("--------------------------")

# Gradient Descent Iterations
for i in range(iterations):

    gradient = derivative(x)

    x = x - learning_rate * gradient

    y = function(x)

    x_points.append(x)
    y_points.append(y)

    print("Iteration:", i+1,
          "X =", round(x, 4),
          "Y =", round(y, 4))

# Final Minimum
print("\nMinimum Point:")
print("X =", round(x, 4))
print("Y =", round(function(x), 4))

# Plot Function
x_values = np.linspace(-6, 4, 100)
y_values = function(x_values)

plt.figure(figsize=(9, 5))

# Function Curve
plt.plot(x_values, y_values,
         label='y = (x+3)^2')

# Gradient Descent Path
plt.plot(x_points, y_points,
         'ro--', label='Gradient Descent Path')

# Starting Point
plt.scatter(2, 25, color='green',
            s=100, label='Starting Point (2,25)')

# Minimum Point
plt.scatter(-3, 0, color='blue',
            s=100, label='Minimum Point (-3,0)')

plt.title("Gradient Descent Algorithm")
plt.xlabel("X Values")
plt.ylabel("Y Values")

plt.legend()
plt.grid(True)

plt.show()
```
