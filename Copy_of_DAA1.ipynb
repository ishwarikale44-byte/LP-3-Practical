```python
# Practical 1: Fibonacci Numbers and Step Count

def fibonacci_with_step_count(n):

    a = 0
    b = 1
    steps = 0
    fibonacci_series = []

    return a, b, steps, fibonacci_series
```

```python
# Function to generate Fibonacci series

def generate_series(n):

    a, b, steps, fibonacci_series = fibonacci_with_step_count(n)

    for i in range(n):

        fibonacci_series.append(a)

        next_number = a + b

        a = b
        b = next_number

        steps = steps + 1

    return fibonacci_series, steps
```

```python
# Accept input from the user

n = int(input("Enter the number of Fibonacci terms: "))

# Check whether the input is valid
if n <= 0:
    print("Please enter a positive number of terms.")

else:
    series, steps = generate_series(n)

    print("\nFibonacci Series:", series)
    print("Total Number of Terms:", n)
    print("Total Step Count:", steps)
```
