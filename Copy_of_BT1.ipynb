```python
print("META MASK WALLET")
print("Ether Transaction Fee Calculation")
```

```python
# Initialize gas used and gas price

gas_used = 21000
gas_price = 10

print("Gas Used:", gas_used)
print("Gas Price:", gas_price, "Gwei")
```

```python
# Calculate transaction fee

transaction_fee = gas_used * gas_price

print("Transaction Fee:", transaction_fee, "Gwei")
```

```python
# Convert Gwei into Ether

ether_fee = transaction_fee / 1000000000

print("Ether Spent:", ether_fee, "ETH")
```

```python
# Display final transaction details

print("----- TRANSACTION DETAILS -----")
print("Cryptocurrency: Ether")
print("Gas Used:", gas_used)
print("Gas Price:", gas_price, "Gwei")
print("Total Fee:", ether_fee, "ETH")
print("Transaction Status: Calculated")
```
