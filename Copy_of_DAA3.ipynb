```python
# Number of items
n = 4

# Weights of items
weights = [2, 3, 4, 5]

# Profits of items
profits = [3, 4, 5, 6]

# Maximum capacity of knapsack
W = 5

print("Number of Items:", n)
print("Weights:", weights)
print("Profits:", profits)
print("Knapsack Capacity:", W)
```

```python
# Create DP table
dp = [[0 for x in range(W + 1)]
      for x in range(n + 1)]

print("DP Table Created Successfully!")
```

```python
# Apply Dynamic Programming

for i in range(1, n + 1):

    for w in range(1, W + 1):

        if weights[i - 1] <= w:

            dp[i][w] = max(
                profits[i - 1] + dp[i - 1][w - weights[i - 1]],
                dp[i - 1][w]
            )

        else:

            dp[i][w] = dp[i - 1][w]

print("Maximum Profit:", dp[n][W])
```
