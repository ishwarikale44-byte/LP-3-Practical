```python
import heapq
```

```python
class Node:
    def __init__(self, char, freq):
        self.char = char
        self.freq = freq
        self.left = None
        self.right = None

    def __lt__(self, other):
        return self.freq < other.freq
```

```python
def huffman_encoding(characters, frequencies):

    heap = []

    for char, freq in zip(characters, frequencies):
        node = Node(char, freq)
        heapq.heappush(heap, node)

    while len(heap) > 1:

        left = heapq.heappop(heap)
        right = heapq.heappop(heap)

        new_node = Node(None, left.freq + right.freq)

        new_node.left = left
        new_node.right = right

        heapq.heappush(heap, new_node)

    return heap[0]
```

```python
def generate_codes(root, code, huffman_codes):

    if root is None:
        return

    if root.char is not None:
        huffman_codes[root.char] = code if code else "0"
        return

    generate_codes(root.left, code + "0", huffman_codes)
    generate_codes(root.right, code + "1", huffman_codes)
```

```python
characters = ['A', 'B', 'C', 'D', 'E', 'F']

frequencies = [5, 9, 12, 13, 16, 45]

root = huffman_encoding(characters, frequencies)

huffman_codes = {}

generate_codes(root, "", huffman_codes)

print("Huffman Encoding Results")
print("------------------------")

for char in characters:
    print(char, ":", huffman_codes[char])
```
