```python
# Mini Project: E-Voting System

class EVoting:

    def __init__(self):
        self.candidates = {
            1: "Candidate A",
            2: "Candidate B",
            3: "Candidate C"
        }

        self.votes = {
            1: 0,
            2: 0,
            3: 0
        }

        self.voters = set()

    # Display candidates
    def display_candidates(self):
        print("\nAvailable Candidates")
        print("----------------------")

        for key, name in self.candidates.items():
            print(key, ".", name)

    # Cast vote
    def vote(self, voter_id, candidate_id):

        if voter_id in self.voters:
            print("You have already voted!")
            return

        if candidate_id not in self.candidates:
            print("Invalid candidate ID!")
            return

        self.voters.add(voter_id)
        self.votes[candidate_id] += 1

        print("Vote successfully recorded!")

    # Display results
    def display_results(self):
        print("\nVoting Results")
        print("----------------------")

        for key, name in self.candidates.items():
            print(name, ":", self.votes[key], "votes")

        print("----------------------")
        print("Total Votes:", sum(self.votes.values()))


# Create voting system
voting = EVoting()

print("WELCOME TO E-VOTING SYSTEM")

# Display candidates
voting.display_candidates()

# Voting process
voting.vote("Voter1", 1)
voting.vote("Voter2", 2)
voting.vote("Voter3", 1)
voting.vote("Voter4", 3)

# Duplicate voting test
voting.vote("Voter1", 2)

# Display final results
voting.display_results()
```
